## Sets

#### Exercise 1:

Write a `Function` that takes `two sets` as input and returns a `set` containing the common elements.

```python
set_1 = {1, 2, 3, 4}
set_2 = {3, 4, 5, 6}

# Add your code here
# a.intersection(b) for commons
def set_comparison(set1, set2):
    answer = set1.intersection(set2)
    return answer

set_comparison(set_1, set_2)
# Expected output: {3, 4}
```

#### Exercise 3:

Write a `Function` that takes a `set` and an `element`. If the element is `NOT` in the set, `add` it and return the updated set.

```python
set_1 = {1, 2, 3}
x = 4

# Add your code here
# set_variable.add(elem)
def inclusion_check(set, element):
    if element not in set:
        set.add(element)
    return set

inclusion_check(set_1, x)

#Expected output: {1, 2, 3, 4}
```

#### Exercise 4:

Write a `Function` that takes `two sets` and returns the `symmetric` difference.

```python
set_1 = {1, 2, 3}
set_2 = {3, 4, 5}

# Add your code here
# a.symmetric_difference(b)
def sym_comp(set_a, set_b):
    sym_comparison = set_a.symmetric_difference(set_b)
    return sym_comparison

sym_comp(set_1, set_2)

# Expected output: {1, 2, 4, 5}
```

#### Exercise 5:

Write a `Function` that takes a `list of sets` and returns their union.

```python
sets = [{1, 2}, {2, 3}, {3, 4}]

# Add your code here
# a.union(b)
def union_comp(set_to_check):
    set_list = list(set_to_check)
    first_set = set(set_list[0])
    sec_set = set(set_list[1])
    third_set = set(set_list[2])
    first_sec_set = first_set.union(sec_set)
    complete_set = first_sec_set.union(third_set)
    return complete_set
  

print(union_comp(sets))
# Expected output:  {1, 2, 3, 4}
```

#### Exercise 4:

Write a `Function` that `merges two` dictionaries and returns the `combined dictionary`.

```python
dict_1 = {"name" : "Alice", "age": 25, "city" : "New York"}
dict_2 = {"favorite food" : "fish tacos", "address" : "123 oak dr"}

# Add your code here
def dict_combo(dictA, dictB):
    dictA.update(dictB)
    return dictA

print(dict_combo(dict_1, dict_2))


# Expected output: {"name" : "Alice", "age": 25, "city" : "New York", "favorite food" : "fish tacos"}
```

```python
dict_1 = {"name" : "Alice", "age" : 25} # Update age to 26

dict_2 = {"name" : "Brandon", "age" : 29} # Attempt to update city to "Houston"

# Add your code here
def specific_key(dictC, key, value):
    if dictC.get(key):
        dictC[key] = [value]
        return dictC
    else:
        print("Key not found")
        return

print(specific_key(dict_1, "age", 26))
print(specific_key(dict_2, "city", "houston"))
        


 
# Expected output: 
#                   {"name" : "Alice", "age" : 26}
#                  "Key not found"
```
