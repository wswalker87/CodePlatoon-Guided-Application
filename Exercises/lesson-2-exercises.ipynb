#### Exercise 2:

Write a `function` that takes a `string` and returns the `number of vowels`.

```python
my_word = "hello world" 

# Add your code here
def vowel_count(string1):
    vowel_sum = 0
    string1.count("a") == vowel_sum
    vowel_sum += string1.count("e")
    vowel_sum += string1.count("i")
    vowel_sum += string1.count("o")
    vowel_sum += string1.count("u")
    print(f"The total number of vowels is {vowel_sum}")
    return vowel_sum

    
vowel_count(my_word)

# Expected output: 3
```

#### Exercise 3:

Write a `function` that converts a temperature from `Celcius to Fahrenheit` and vice versa. The function should take two arguments: The `temperature value` and the `converstion type` ("C to F" or "F to C")

Converstion Formula for `C to F` ---> F = C * (9 / 5) + 32

Converstion Formula for `F to C` ---> C = (F - 32) * 5 / 9

```python
temp_1 = 25 # Temperature in Celcius
temp_2 = 77 # Temperature in Fahrenheit
# Add your code here
def temp_convert(tempvalue,conversiontype):
    if conversiontype == "C to F":
        c_to_f = (tempvalue * (9/5)) + 32
        return c_to_f
    elif conversiontype == "F to C":
        f_to_c = (tempvalue - 32) * (5/9)
        return f_to_c
    else:
        print("You didn't give enough information")


temp_convert(temp_1, "C to F")
temp_convert(temp_2, "F to C")
""" Expected Output:
    77.0
    25.0
"""
```
