#### Exercise 5:

Write a `Function` that checks if a list is sorted in ascending or descending order. Return 'Ascending', 'Descending', or 'Unsorted'.

```python
list_1 = [1, 2, 3]
list_2 = [3, 2, 1]
list_3 = [1, 3, 2]

# Add your code here
def asc_or_desc(listicle):
    if listicle[0] < listicle[1] and listicle[1] < listicle[2]:
        list_is = "Ascending"
    elif listicle[0] > listicle[1] and listicle[1] > listicle[2]:
        list_is = "Descending"
    else:
        list_is = "Unsorted"
    return list_is


print(asc_or_desc(list_3))

""" 
Expected output:
    Ascending
    Descending
    Unsorted

"""
```

#### Exercise 1:

Write a `Function` that returns the `index` of a `value` in a `tuple`. If the value is not present, return -1.

```python
tuple_1 = (10, 20, 30) # Look for 20 the first time, and then look for 40

# Add your code here
def find_index(tuple, value):
    if tuple.count(value):
        found_index = tuple.index(value)
        return found_index
    else:
        no_match = -1
        return no_match
    
print(find_index(tuple_1, 40))

""" 
Expected output:
    1
    -1
"""
```

#### Exercise 3:

Write a `Function` that swaps the `first` and `last` elements of a `tuple`. If the tuple has fewer than two elements, return the original tuple.

```python
tuple_1 = (1, 2, 3, 4)
tuple_2 = (1,)

# Add your code here
def swapper(tup):
    if len(tup) < 2:
        return tup
    else:
        tup_list = list(tup)
        last_value = tup_list[-1]
        first_value = tup_list[0]
        tup_list.pop(0)
        tup_list.pop(-1)
        tup_list.insert(0, last_value) == tup_list
        tup_list.append(first_value)
        once_again_tup = tuple(tup_list)
        return once_again_tup

print(swapper(tuple_1))
""" 
Expected output:
    (4, 2, 3, 1)
    (1,)
"""
```

#### Exercise 4:

Write a `Function` that creates a `tuple` with a given element repeated `n` times.

```python
tuple_1 = (5,) # repeat it by 3
tuple_2 = ('a',) # repeat it by 0

# Add your code here
def repeat_element(tup, n):
    new_tup = (tup * n)
    return new_tup

print(repeat_element(tuple_1, 3))
print(repeat_element(tuple_2, 0))


""" 
Expected output:
    (5, 5, 5)
    ()
"""
```
